# Matrix Definition

```python
A = np.array([[1.2, 0.1],
              [0.05, 1.1]])
A
```

Explanation:

This matrix represents a linear transformation with a slight stretching along one axis compared to the perpendicular axis.

# Eigenvalues & Eigenvectors of Matrix A

```python
A = np.array([[1.2, 0.1],
              [0.05, 1.1]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

```

Explanation:

Eigenvalues show how much stretching happens in special directions.

1.2366 - strong stretch direction

1.0634 - mild stretch direction

Eigenvectors represent the direction which does not change during the linear transformation.

# Exponential Growth Model at t = 20

```python
t = 20
growth_rate = 30 * np.exp(0.03 * t)

print("Growth rate at t = 20:", growth_rate)
```

Explanation:

In 20 units of time, the quantity increases from 30 to 54.66.
Typical growth curve: slow start, then gradually accelerates.

# Exponential Growth Model (Graphical Presentation)

```python
t = np.linspace(0, 20, 200)
growth_rate = 30 * np.exp(0.03 * t)

t0 = 20
g0 = 30 * np.exp(0.03 * t0)

plt.plot(t, growth_rate)
plt.scatter([t0], [g0])
plt.title("Growth Rate Over Time")
plt.xlabel("t (years)")
plt.ylabel("Growth rate")
plt.grid(True)
plt.show()
plt.savefig("math_1.png",dpi=1000)
```

# Poisson Distribution (λ = 5)

```python
lam = 5
print("Expected Value:", stats.poisson.mean(lam))
print("Variance:", stats.poisson.var(lam))
```

Explanation:

For Poisson distributions, mean = variance = λ.
In this case, we have an average of 5 events, with equal variance.

# Poisson Distribution (Graphical Presentation)

```python
lam = 5
k = np.arange(0, 16)
pmf = stats.poisson.pmf(k, lam)

plt.bar(k, pmf)
plt.title("Poisson Distribution PMF (λ = 5)")
plt.xlabel("Number of immigration events (k)")
plt.ylabel("P(X = k)")
plt.show()
plt.savefig("math_2.png",dpi=1000)

```

# 95% Confidence Interval for Population Mean

```python
data = np.array([1050, 1070, 1065, 1080, 1045, 1075, 1060, 1070])

mean = np.mean(data)
std = np.std(data, ddof=1)
ci = stats.t.interval(0.95, len(data)-1, loc=mean, scale=std/np.sqrt(len(data)))

print("Mean:", mean)
print("Standard Deviation:", std)
print("95% Confidence Interval:", ci)
```

Explanation:

“We are 95% confident the true population mean lies between 1054.27 and 1074.48.”

Because sample size is small → t-distribution used.

# 95% Confidence Interval for Population Mean(Graphical Presentation)


```python
data = np.array([1050, 1070, 1065, 1080, 1045, 1075, 1060, 1070])
mean = np.mean(data)
std = np.std(data, ddof=1)

ci_low, ci_high = stats.t.interval(0.95, len(data)-1, loc=mean, scale=std/np.sqrt(len(data)))

plt.errorbar(x=1, y=mean, yerr=[[mean-ci_low], [ci_high-mean]], fmt='o', capsize=8)
plt.title("95% Confidence Interval for True Mean")
plt.xticks([1], ["District mean"])
plt.ylabel("Population")
plt.grid(True)
plt.show()
```

```python
data = np.array([1050, 1070, 1065, 1080, 1045, 1075, 1060, 1070])
mean = np.mean(data)

plt.hist(data, bins=5, edgecolor="black")
plt.axvline(mean, linestyle="dashed", linewidth=2, label=f"Mean = {mean:.3f}")
plt.title("District Populations (Histogram)")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
